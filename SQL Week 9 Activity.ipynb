```python
# Import necessary libraries if not already imported
import pandas as pd
import sqlite3

# Drop the existing table before inserting new data
# This ensures a clean slate and avoids the "table already exists" error
try:
    conn.execute("DROP TABLE IF EXISTS Listings")
    conn.commit()  # Commit the drop operation
except:
    pass  # If there's any issue dropping, continue anyway

# Now read the CSV and insert into the database
pd.read_csv('realestate_listings.csv').to_sql('Listings', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT *
FROM listings
ORDER BY sqft DESC LIMIT 3
''', conn)
```

**Task 4**

```python
import pandas as pd

# Drop the existing table if it exists
try:
    conn.execute("DROP TABLE IF EXISTS Campaigns")
    conn.commit()  # Commit the drop operation
except:
    pass  # If table doesn't exist or error occurs, continue

# Now read the CSV and create the table
pd.read_csv('marketing_campaigns.csv').to_sql('Campaigns', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT *
FROM campaigns
GROUP BY channel
''', conn)
```

```python
pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY week
''', conn)
```

**Task 5**

```python
import sqlite3

# Drop the table if it exists (handles case-sensitivity issues)
cursor = conn.cursor()
cursor.execute("DROP TABLE IF EXISTS Orders")
conn.commit()

# Now insert the data
pd.read_csv('restaurant_orders.csv').to_sql('Orders', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT order_date, ROUND(SUM(order_total), 2) AS total_orders
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

```python
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT doctor,
       ROUND(AVG(patient_age), 1) AS patient_age,
       MAX(wait_minutes) AS wait_minutes
FROM visits
GROUP BY doctor
''', conn)
```

**Task 7**

```python
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT c.branch,
SUM(t.amount) AS total_transaction_amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
